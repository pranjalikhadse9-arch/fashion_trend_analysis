```python
#Boxplot
```

```python
sb.boxplot(df.Original_Price)
plt.show()
```

```python
i = 'Original_Price'

q1 = np.percentile(df[i],25)
q3 = np.percentile(df[i],75)

iqr = q3 - q1
```

```python
c1 = q1 - 1.5 * (iqr)
c2 = q3 + 1.5 * (iqr)
```

```python
t = df[(df[i]>c2) | (df[i]<c1)]
t = t.index
```

```python
df = df.drop(labels=t)
```

```python
sb.boxplot(df.Discount_Price)
plt.show()
```

```python
i = 'Discount_Price'

q1 = np.percentile(df[i],25)
q3 = np.percentile(df[i],75)

iqr = q3 - q1
```

```python
c1 = q1 - 1.5 * (iqr)
c2 = q3 + 1.5 * (iqr)
```

```python
t = df[(df[i]>c2) | (df[i]<c1)]
t = t.index
```

```python
df = df.drop(labels=t)
```
